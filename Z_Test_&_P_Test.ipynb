```python
import numpy as np
```

```python
data = [52,48,51,49,50,52,51,54,50]
```

```python
# Known Population mean
population_men = 50 #(u)
```

```python
# know population std. deviation
population_std = 5
```

```python
# Sample mean ]
sample_men = np.mean(data)
```

```python
# significance level
alpha = 0.5
```

```python
z_score = (sample_men - population_men) / (population_std / np.sqrt(len(data)))
```

```python
z_score
```

```python
from scipy.stats import norm
p_value = 2*(1-norm.cdf(abs(z_score)))
#normal distridution(norm)
#comulative distribution function
```

```python
p_value
```

```python
if p_value<alpha:
  print("Reject the null hypothesis")
else:
  print("Fail to reject the null hypothesis")
```

```python
from scipy.stats import ttest_1samp
```

```python
# sample data
data = [145,148,142,150,144,147,143,145,149,141]
```

```python
# Population mean
population_mean = 150
```

```python
t_value, p_values = ttest_1samp(data,population_mean)
```

```python
t_value
```

```python
p_values
```

```python
if p_values < 0.05:
  print("Reject the null hypothesis")
else:
  print("Fail to reject the null hypothesis")
```

```python
import pandas as pd
contigency_table=pd.crosstab(df["class"],df["survived"])
```

```python
contigency_table
```

```python
from scipy.stats import chi2_contingency
```

```python
chi2,p,dof,expected=chi2_contingency(contigency_table)
```

```python
print(chi2)
print(p)
print(dof)
print(expected)
```

```python
if chi2 > p:
  print("Reject the null hypothesis")
else:
  print("Fail to reject the null hypothesis")
```
