```python
fertlizer_A   = [20,22,19,21,23]
fertilizer_B  = [25,27,24,26,28]
fertilizer_C  = [30,32,29,31,33]
```

```python
alpha = 0.05
```

```python
#1. Mean of each Groups
import numpy as np
mean_A = np.mean(fertlizer_A)
mean_B = np.mean(fertilizer_B)
mean_C = np.mean(fertilizer_C)

all_data = fertlizer_A + fertilizer_B + fertilizer_C

overall_mean = np.mean(all_data)

n = len(fertlizer_A)
```

```python
SSB =  n*((mean_A - overall_mean)**2 +(mean_B-overall_mean)**2+ (mean_C-overall_mean)**2)
```

```python
SSB
```

```python
SSW = (sum((i - mean_A)**2 for i in fertlizer_A)+
      sum((i - mean_B)**2 for i in fertilizer_B)+
      sum((i - mean_C)**2 for i in fertilizer_C))
```

```python
SSW
```

```python
k = 3
N = len(all_data)
```

```python
dfbetween = k-1
dfwithin = N-k
```

```python
print(dfbetween)
print(dfwithin)
```

```python
MSB = SSB/dfbetween
```

```python
MSW = SSW/dfwithin
```

```python
F_calculated = MSB / MSW
```

```python
F_calculated
```

```python
from scipy import stats
F_critical = stats.f.ppf(1-alpha, dfbetween, dfwithin)
```

```python
F_critical
```

```python
if F_calculated > F_critical:
  print("We Reject Ho there is significat difference between groups")
else:
  print("Fail to reject Ho: no significat difference")
```

```python
F_value , p_value = stats.f_oneway(fertlizer_A,fertilizer_B, fertilizer_C)
```

```python
if p_value < alpha:
  print("We Reject Ho there is significat difference between groups")
else:
  print("Fail to reject Ho: no significat difference")
```
