```python
 import numpy as np
```

```python
data=[12,13,14,16,18,19,20,25,27,29,30,34,35,38,40,210,-220]
```

```python
import seaborn as sns
```

```python
sns.boxplot(x=data)
```

```python
Q1=np.percentile(data,25)
Q3=np.percentile(data,75)
```

```python
Q1
#Q3
```

```python
Q3
```

```python
IQR=Q3-Q1
```

```python
uf=1.5*IQR+Q3
lf=1.5*IQR-Q1
```

```python
clean_data=[]
```

```python
for i in data :


  if i > uf :
    continue


  elif i < lf :
    continue


  else :
    clean_data.append(i)
print(clean_data)


```
